### Se quiere simular el flujo del gradiente en un grafo incompleto para encontrar sus twisted states estables

```python
import numpy as np
```

```python
n = 10 # cantidad de vértices del grafo
q = 1

theta = np.zeros(n)
for v in range(n): # tengo que inicializar las fases para que estén torcidas
    theta[v] = 2 * np.pi * q * v / n

w = np.zeros((n,n)) # matriz de adyacencia para un grafo cíclico
for i in range(n): # solo están conectados los que están inmediatamente al lado
    w[i, (i + 1) % n] = 1
    w[i, (i - 1) % n] = 1
print(w)
```

```python
def velocidades (n,theta,w):
    velocidades = np.array(np.zeros(n))
    for i in range(n): # agarro un nodo
        velocidad = 0
        for j in range(n): # lo hago interactuar con todos los demás
            if i != j:
                velocidad += w[i,j] * np.sin(theta[j]-theta[i])
        velocidades[i] = velocidad
    return velocidades

print(velocidades(n,theta,w))
```

```python
# Arranco con la simulación
def simular(n,theta,w,secs,dt=0.01):
    theta = theta.copy()
    iteraciones = int(secs * 1/dt)
    posiciones = np.zeros((iteraciones,n)) # serían todos los thetas en cada instante
    for i in range(iteraciones):
        vel = velocidades(n,theta,w)
        nueva_theta = theta + vel * dt
        posiciones[i] = nueva_theta
        theta = nueva_theta
    return posiciones # me termina quedando una matriz en la que cada fila son las fases cada 0.01 segundos

posiciones = simular(n,theta,w,5)
print(posiciones)
```

```python
# veo si se sincronizaron con el parámetro de orden
r = np.mean(np.exp(1j * posiciones), axis=1)
r_abs = np.abs(r)
print(r_abs)
```

Como se puede ver, r da practicamente cero, así que este grafo no sincroniza.
