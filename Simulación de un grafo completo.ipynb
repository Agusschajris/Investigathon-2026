### Se quiere simular el flujo del gradiente en un grafo completo para verificar que siempre sincroniza

```python
import numpy as np
```

```python
n = 5 # cantidad de vértices del grafo
theta = np.random.uniform(-np.pi, np.pi, n) # asigno fases aleatorias entre -pi y pi
w = 1 # matriz de adyacencia para que todos los vértices estén conectados
```

```python
def velocidades (n,theta,w):
    velocidades = np.array(np.zeros(n))
    for i in range(n): # agarro un nodo
        velocidad = 0
        for j in range(n): # lo hago interactuar con todos los demás
            if i != j:
                velocidad += w * np.sin(theta[j]-theta[i])
        velocidades[i] = velocidad
    return velocidades

print(velocidades(n,theta,w))
```

```python
# Arranco con la simulación
def simular(n,theta,w,secs,dt=0.01):
    theta = theta.copy()
    iteraciones = int(secs * 1/dt)
    posiciones = np.zeros((iteraciones,n)) # serían todos los thetas en cada instante
    for i in range(iteraciones):
        vel = velocidades(n,theta,w)
        nueva_theta = theta + vel * dt
        posiciones[i] = nueva_theta
        theta = nueva_theta
    return posiciones # me termina quedando una matriz en la que cada fila son las fases cada 0.01 segundos

posiciones = simular(n,theta,w,5)
print(posiciones)
```

```python
# veo si se sincronizaron con el parámetro de orden
r = np.mean(np.exp(1j * posiciones), axis=1)
r_abs = np.abs(r)
print(r_abs)
```

Se puede ver que el grafo se termina sincronizando casi enseguida, dado que los últimos parámetros de orden en el tiempo son practicamente 1
